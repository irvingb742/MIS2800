## YOUR ANSWERS TO PART 2 BELOW

Question 1:
In Redshift architecture, columnar storage is often more important to analysts than perfectly normalized tables, since they typically analyze specific columns rather than entire rows. Columnar storage organizes data by columns rather than rows, allowing queries to scan only the necessary data.
Question 2:
The Leader Node and Compute Nodes setup speeds up Business Intelligence reporting by splitting the work across several systems. The Leader Node gets the query, makes a plan, and sends tasks to the Compute Nodes. Each Compute Node then processes its part of the data at the same time. Since queries run at the same time instead of one after another, reports are created much faster. This way of processing lets businesses analyze large datasets quickly and get insights in seconds or minutes instead of hours.

Question 1:
As a future business leader, learning tools like Redshift would change how I talk with my data team. Instead of just asking what happened, I could dig deeper by asking why trends happened, what patterns might predict the future, and how data can help guide strategic decisions. Understanding what a data warehouse can do also helps leaders ask for real-time dashboards, predictive analysis, and comparisons of performance across seasons or markets. Question 2:
The “Clean and Load” phase of the ETL process is the most important step in a Business Intelligence strategy because bad data can cause wrong insights. In this phase, data is fixed, standardized, and checked before it goes into the data warehouse. If errors, duplicates, or missing values aren’t fixed, reports and analytics might give misleading results, which can lead to bad business decisions. Clean data guarantees accuracy, consistency, and trust in dashboards and reports, forming the base for good analytics and decision-making.
