## YOUR ANSWERS TO PART 1 BELOW

Question 1:
Switching from simple data storage to the Amazon Redshift data warehouse let Lyft analyze large amounts of past ride data instead of just tracking single rides. Using Redshift, Lyft combined data from millions of trips to spot patterns in rider behavior, popular locations, and travel times. This shift helped the company move beyond basic record keeping to data-driven innovation. For example, Lyft Line was developed by studying ride patterns and finding when multiple riders were heading in similar directions. The data warehouse helped Lyft turn stored data into insights that supported new products and business plans.

Question 2:
Lyft cannot use the same database for OLTP and OLAP because they serve different functions. OLTP systems handle fast, real-time transactions like booking rides, updating driver locations, and processing payments, requiring instant responses. OLAP systems, such as Redshift, analyze large datasets and run complex queries for planning and decision-making. Combining both would slow real-time app performance due to the high processing demands of analytical queries.

## YOUR ANSWERS TO PART 2 BELOW

Question 1:
Columnar storage matters more to analysts than perfectly normalized tables because analytical queries usually target specific columns instead of whole rows. By organizing data by columns, Redshift can scan just the information it needs when running queries.

Question 2:
The Leader Node and Compute Nodes architecture speeds up Business Intelligence reports by distributing work across multiple machines. The Leader Node receives queries and assigns tasks to Compute Nodes, which process data concurrently. This parallel processing enables faster analysis of large datasets than a single system could achieve.
